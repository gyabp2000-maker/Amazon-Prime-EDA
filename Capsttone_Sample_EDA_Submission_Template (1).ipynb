## Handling Missing Values

````python
# Write your code to make your dataset analysis ready.
# Missing values percentage

missing_values = (titles.isnull().sum() / len(titles)) * 100

missing_df = pd.DataFrame({
    'Column': missing_values.index,
    'Missing Percentage': missing_values.values
})

missing_df.sort_values(by='Missing Percentage', ascending=False)
````

````python
# Filling numerical missing values with median

titles['imdb_score'].fillna(titles['imdb_score'].median(), inplace=True)
titles['imdb_votes'].fillna(titles['imdb_votes'].median(), inplace=True)
titles['tmdb_popularity'].fillna(titles['tmdb_popularity'].median(), inplace=True)
titles['tmdb_score'].fillna(titles['tmdb_score'].median(), inplace=True)
titles['description'].fillna('Description Not Available', inplace=True)
titles['imdb_id'].fillna('Unknown', inplace=True)
````

````python
# Filling categorical missing values

titles['age_certification'].fillna('Unknown', inplace=True)
titles['seasons'].fillna(0, inplace=True)
````

````python
titles.isnull().sum()
````

````python
credits.isnull().sum()
````

````python
# Filling missing character names

credits['character'].fillna('Unknown Character', inplace=True)
````

````python
# Final missing values check

credits.isnull().sum()
````

## Feature Engineering

Feature engineering helps create new meaningful variables that improve analysis quality.

In this project:
- Main genre will be extracted
- Runtime categories will be created
- Decade-based analysis can be enabled

###Extract Main Genre

````python
# Extracting primary genre
titles['main_genre'] = (
    titles['genres']
    .astype(str)
    .str.strip('[]')
    .str.replace("'", "", regex=False)
    .str.split(',')
    .str[0]
    .str.strip()
)
````

###Runtime Category Creation

````python
# Creating runtime categories

def runtime_category(x):
    if x < 60:
        return 'Short'
    elif x < 120:
        return 'Medium'
    else:
        return 'Long'

titles['runtime_category'] = titles['runtime'].apply(runtime_category)
````

###Creating Decade Column

````python
# Creating decade column

titles['decade'] = (titles['release_year'] // 10) * 10
````
